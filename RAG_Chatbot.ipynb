```python
from langchain_community.document_loaders import PyPDFLoader
import os
pdf_path = list(uploaded.keys())[0]
loader = PyPDFLoader(pdf_path)
documents = loader.load()
print("Total pages:", len(documents))
print("\nFirst page content:\n")
print(documents[0].page_content)
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)
chunks = text_splitter.split_documents(documents)
print("Total chunks:", len(chunks))
```

```python
for i, chunk in enumerate(chunks[:5]):
    print(f"\n--- CHUNK {i+1} ---")
    print(chunk.page_content)
    print("Metadata:", chunk.metadata)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

```python
text = "Machine learning is a subset of artificial intelligence."

vector = embedding_model.embed_query(text)

print(vector)
print("Vector dimensions:", len(vector))
```

```python
texts = [chunk.page_content for chunk in chunks]

vectors = embedding_model.embed_documents(texts)

print("Total chunks:", len(chunks))
print("Total embeddings:", len(vectors))
print("Dimensions of each embedding:", len(vectors[0]))
```

```python
import numpy as np

A = np.array([1, 2, 3])
B = np.array([2, 4, 6])

similarity = np.dot(A, B) / (
    np.linalg.norm(A) * np.linalg.norm(B)
)

print("Cosine Similarity:", similarity)
```

```python
from langchain_chroma import Chroma

vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embedding_model,
    collection_name="pdf_rag_collection"
)

print("Documents stored successfully!")
```

```python
query = "Explain the main concepts discussed in this document."

results = vectorstore.similarity_search(
    query,
    k=3
)

for i, doc in enumerate(results):
    print(f"\n--- RESULT {i+1} ---")
    print(doc.page_content)
    print("Metadata:", doc.metadata)
```

```python
from google.colab import userdata
from langchain_google_genai import ChatGoogleGenerativeAI

google_api_key = userdata.get("GOOGLE_API_KEY")

llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    google_api_key=google_api_key,
    temperature=0
)
```

```python
response = llm.invoke("Explain machine learning in 3 lines.")
print(response.content)
```

```python
query = "What are the main topics discussed in this PDF?"

retrieved_docs = vectorstore.similarity_search(
    query,
    k=3
)

for doc in retrieved_docs:
    print(doc.page_content)
    print("----------------")
```

```python
context = "\n\n".join(
    [doc.page_content for doc in retrieved_docs]
)

print(context)
```

```python
from langchain_core.prompts import ChatPromptTemplate

prompt = ChatPromptTemplate.from_template("""
You are a helpful AI assistant.

Answer the question using ONLY the provided context.

If the answer is not present in the context,
clearly say that the information is not available
in the uploaded document.

Context:
{context}

Question:
{question}

Answer:
""")
```

```python
chain = prompt | llm
```

```python
response = chain.invoke({
    "context": context,
    "question": query
})

print(response.content)
```

```python
def ask_pdf(question):

    # Step 1: Retrieve relevant documents
    retrieved_docs = vectorstore.similarity_search(
        question,
        k=3
    )

    # Step 2: Convert documents into context
    context = "\n\n".join(
        [doc.page_content for doc in retrieved_docs]
    )

    # Step 3: Generate answer
    response = chain.invoke({
        "context": context,
        "question": question
    })

    # Step 4: Return answer and sources
    return {
        "answer": response.content,
        "sources": retrieved_docs
    }
```

```python
result = ask_pdf(
    "Explain the important concepts in this document."
)

print("ANSWER:")
print(result["answer"])

print("\nSOURCES:")

for doc in result["sources"]:
    print(doc.metadata)
```
